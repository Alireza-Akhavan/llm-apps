````python
prompt = "یک پاراگراف در مورد ایران بنویس"

response = completion(
    model="ollama/gemma3:4b",
    messages=[{"role": "user", "content": prompt}],
    stream=True  # فعال کردن استریم
)

# چاپ استریمی
for chunk in response:
    if chunk['choices'][0].get('delta', {}).get('content'):
        print(chunk['choices'][0]['delta']['content'], end='', flush=True)
````

## Image Analysis with LiteLLM (Vision Support)

````python
from litellm import completion
import base64

def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')

image_path = "images/image.webp"
base64_image = encode_image(image_path)

response = completion(
    model="openai/gpt-4o",  # or "ollama/llava" or ... for local vision model
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "یه اسم زیبا برای تصویر زیر بزار"},
                {
                    "type": "image_url",
                    "image_url": {
                        "url": f"data:image/jpeg;base64,{base64_image}"
                    }
                }
            ]
        }
    ]
)

print(response['choices'][0]['message']['content'])
````
