## Model Limitations: Hallucinations


```python
import litellm
from litellm import completion
from IPython.display import display, Markdown

def get_completion(prompt, model="ollama/llama3.1:8b", temperature=0.2):
    messages = [{"role": "user", "content": prompt}]
    
    response = completion(
        model=model,
        messages=messages,
        temperature=temperature,
    )
    
    return response.choices[0].message.content

def ltr_print(text):
    display(Markdown(text))
    
def rtl_print(text):
    display(Markdown(f'<span style="direction: rtl; display: block; text-align: right;">{text}</span>'))
```

```python
prompt = f"""
یک جمله که قبلا علی شریعتی دقیقا گفته  پیرامون سرمایه گذاری در بورس طلا در شرایط جنگی را بازگو کن"""
response = get_completion(prompt)
rtl_print(response)
```

```python
prompt = f"""
در مورد مسواک های برقی شرکت سایپا بهم اطلاعات بده.
"""
response = get_completion(prompt)
rtl_print(response)
```

```python
prompt = f"""
در مورد تلفن همراه های ساخت شرکت ایران خودرو بهم اطلاعاتی بده.
"""
response = get_completion(prompt,temperature=0.2)
rtl_print(response)
```

```python
prompt = f"""
به سوال تا حد امکان صادقانه پاسخ بده و اگر از پاسخ مطمئن نیستید، بگو :
"ببخشید، نمی دانم."
سوال:
در مورد تلفن همراه های ساخت شرکت ایران خودرو بهم اطلاعاتی بده.

پاسخ:

"""
response = get_completion(prompt)
print(response)
```
