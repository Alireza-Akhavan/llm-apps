# Named Entity Recognition


````python
import litellm
from litellm import completion
from IPython.display import display, Markdown

def get_completion(prompt, model="ollama/gemma4:e4b", temperature=0.2):
    messages = [{"role": "user", "content": prompt}]
    
    response = completion(
        model=model,
        messages=messages,
        temperature=temperature,
    )
    
    return response.choices[0].message.content

def ltr_print(text):
    display(Markdown(text))
    
def rtl_print(text):
    display(Markdown(f'<span style="direction: rtl; display: block; text-align: right;">{text}</span>'))
````

````python
p = """
حسن روحانی در دیدار با روحانیون قم در مورد آیت الله رئیسی با شاهروردی دیدار داشته و صحبت کرد.
"""

prompt = f"""
Your task is to extract all person names.
Original paragraph: ```{p}```
"""
response = get_completion(prompt)
print(response)
````
