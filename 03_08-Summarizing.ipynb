# Summarizing


````python
import litellm
from litellm import completion
from IPython.display import display, Markdown

def get_completion(prompt, model="ollama/gemma3:4b", temperature=0.5):
    messages = [{"role": "user", "content": prompt}]
    
    response = completion(
        model=model,
        messages=messages,
        temperature=temperature,
    )
    
    return response.choices[0].message.content

def ltr_print(text):
    display(Markdown(text))
    
def rtl_print(text):
    display(Markdown(f'<span style="direction: rtl; display: block; text-align: right;">{text}</span>'))
````

## Summarize with a word/sentence/character limit

````python
prompt = f"""
Your task is to generate a short summary of a text in Persian \
in at most 30 words.

summary: ```{p}```
"""

response = get_completion(prompt)
rtl_print(response)
````

## Summarize with a focus on shipping and delivery

````python
prompt = f"""
Your task is to generate a short summary of a product \
review from an ecommerce site to give feedback to the \
Shipping deparmtment. 

Summarize the review below, delimited by triple 
backticks, in at most 30 words, and focusing on any aspects \
that mention shipping and delivery of the product. 

The summary must be in Persian.

Review: ```{prod_review}```
"""

response = get_completion(prompt,model="ollama/gpt-oss:20b")
rtl_print(response)
````
