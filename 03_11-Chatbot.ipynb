# فرمت چت

```python
from litellm import completion
from IPython.display import display, Markdown


def get_completion(prompt, model="ollama/gemma-4-26B"):
    messages = [{"role": "user", "content": prompt}]
    response = completion(
        model=model,
        messages=messages,
        temperature=0,
    )
    return response["choices"][0]["message"]["content"]

def get_completion_from_messages(messages, model="ollama/gemma4:e4b", temperature=0):
    response = completion(
        model=model,
        messages=messages,
        temperature=temperature,
    )
    return response["choices"][0]["message"]["content"]


def ltr_print(text):
    display(Markdown(text))

def rtl_print(text):
    text_with_br = text.replace('\n', '<br>')
    display(Markdown(f'<span style="direction: rtl; display: block; text-align: right;">{text_with_br}</span>'))
```

## مثال ۱ — مکالمه با حافظه

```python
messages = [  {'role': 'user', 'content': 'پایتخت ایران کجاست؟ در یک کلمه بگو'}]

response = get_completion_from_messages(messages, temperature=0.2)
rtl_print(response)
```

# ربات سفارش پیتزا

```python
import panel as pn
pn.extension()

# استایل inline برای هر المان - بدون global CSS
rtl_style = {'direction': 'rtl', 'font-family': 'Vazir, sans-serif'}

panels = []
context = [{'role': 'system', 'content': """
تو یک ربات سفارش‌گیر خودکار برای یک رستوران 'پیتزای ما' هستی.
ابتدا به مشتری خوش‌آمد می‌گویی، سپس سفارش را می‌گیری،
و بعد می‌پرسی که سفارش تحویل است یا حضوری.
صبر می‌کنی تا سفارش کامل شود، سپس آن را خلاصه می‌کنی
و می‌پرسی که آیا چیز دیگری می‌خواهند اضافه کنند.
اگر سفارش تحویل است، آدرس را می‌پرسی.
در نهایت مبلغ را اعلام و پرداخت را دریافت می‌کنی.
حتماً سایز و گزینه‌های هر آیتم را دقیق مشخص کن.
با لحنی کوتاه، دوستانه و محاوره‌ای پاسخ بده.
فقط به فارسی پاسخ بده.
منو رستوران:
پیتزا پپرونی: بزرگ ۶۵۰،۰۰۰ / متوسط ۴۵۰،۰۰۰ / کوچک ۲۸۰،۰۰۰ تومان
پیتزا رست بیف: بزرگ ۷۲۰،۰۰۰ / متوسط ۵۰۰،۰۰۰ / کوچک ۳۲۰،۰۰۰ تومان
پیتزا مرغ و اسفناج: بزرگ ۶۸۰،۰۰۰ / متوسط ۴۷۰،۰۰۰ / کوچک ۳۰۰،۰۰۰ تومان
سیب‌زمینی سرخ‌کرده: بزرگ ۲۲۰،۰۰۰ / کوچک ۱۳۰،۰۰۰ تومان
سالاد یونانی: ۱۸۰،۰۰۰ تومان
نوشیدنی‌ها:
آب معدنی: ۴۰،۰۰۰ تومان
نوشابه فانتا: کوچک ۱۰۰،۰۰۰ / بزرگ ۱۵۰،۰۰۰ تومان
نوشابه کوکا: کوچک ۱۰۰،۰۰۰ / بزرگ ۱۵۰،۰۰۰ تومان
"""}]

greeting = get_completion_from_messages(context + [{'role': 'user', 'content': 'شروع مکالمه را با خوش‌آمدگویی آغاز کن'}])
context.append({'role': 'user', 'content': 'شروع مکالمه را با خوش‌آمدگویی آغاز کن'})
context.append({'role': 'assistant', 'content': greeting})

panels.append(pn.Row(
    pn.pane.Str('ربات:', styles=rtl_style),
    pn.pane.Markdown(greeting, width=600, styles={**rtl_style, 'background-color': '#F6F6F6'}),
    styles=rtl_style
))

inp = pn.widgets.TextInput(placeholder='پیام خود را اینجا بنویسید...', styles=rtl_style)
button_conversation = pn.widgets.Button(name='ارسال', styles=rtl_style)

def collect_messages(_):
    prompt = inp.value
    inp.value = ''
    if not prompt.strip():
        return pn.Column(*panels, styles=rtl_style)
    context.append({'role': 'user', 'content': f"{prompt}"})
    response = get_completion_from_messages(context)
    context.append({'role': 'assistant', 'content': f"{response}"})
    panels.append(pn.Row(
        pn.pane.Str('کاربر:', styles=rtl_style),
        pn.pane.Markdown(prompt, width=600, styles=rtl_style),
        styles=rtl_style
    ))
    panels.append(pn.Row(
        pn.pane.Str('ربات:', styles=rtl_style),
        pn.pane.Markdown(response, width=600, styles={**rtl_style, 'background-color': '#F6F6F6'}),
        styles=rtl_style
    ))
    return pn.Column(*panels, styles=rtl_style)

interactive_conversation = pn.bind(collect_messages, button_conversation)

dashboard = pn.Column(
    inp,
    pn.Row(button_conversation, styles=rtl_style),
    pn.panel(interactive_conversation, loading_indicator=True, height=300),
    styles=rtl_style
)

dashboard.show()
```
