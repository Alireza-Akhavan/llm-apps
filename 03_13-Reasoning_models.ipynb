```python
import time
import litellm
from litellm import completion
from IPython.display import display, Markdown

def get_completion(prompt, model="ollama/gemma4:e4b", temperature=0.2, think=False):
    messages = [{"role": "user", "content": prompt}]
    
    response = completion(
        model=model,
        messages=messages,
        temperature=temperature,
        think=think
    )
    
    return response.choices[0].message.content

def ltr_print(text):
    display(Markdown(text))
    
def rtl_print(text):
    text_with_br = text.replace('\n', '<br>')
    display(Markdown(f'<span style="direction: rtl; display: block; text-align: right;">{text_with_br}</span>'))
```

```python
tik = time.time()
get_completion("سلام. چه طوری؟", think=False)
tok = time.time()
print(tok-tik)
```

```python
tik = time.time()
get_completion("سلام. چه طوری؟", think=True)
tok = time.time()
print(tok-tik)
```

```python
QUESTION = "جواب را در یک کلمه بگو، اگر یک قطار با سرعت 120 کیلومتر در ساعت حرکت کند و بعد از 45 دقیقه توقف کند، چند کیلومتر طی کرده؟"
```

```python
tik = time.time()
r = get_completion(QUESTION, think=False)
tok = time.time()
print(f"{round(tok-tik,1)} s")
print(f"answer:{r}")
```

```python
tik = time.time()
r = get_completion(QUESTION, think=True)
tok = time.time()
print(f"{round(tok-tik,1)} s")
print(f"answer:{r}")
```

```python
prompt = "سلام"
messages = [{"role": "user", "content": prompt}]

response = completion(
    model="ollama/gpt-oss:20b",
    messages=messages,
    temperature=0.2,
    think=True,
        options={
        "temperature": 0.2,
        "reasoning_effort": "low"   # low | medium | high
    }
)
response
```

```python
def test_oss(prompt, level):
    messages = [{"role": "user", "content": prompt}]
    
    response = completion(
        model="ollama/gpt-oss:20b",
        messages=messages,
        temperature=0.2,
            options={
            "reasoning_effort": level   # low | medium | high
        }
    )
    return response.choices[0].message.content
```

```python
QUESTION = "فقط فقط یک  تابع پایتونی بنویس که یک لیست از اعداد را گرفته و در سریعترین زمان ممکنه آنها را سورت کند"
```

```python
tik = time.time()
r = test_oss(QUESTION, 'low')
tok = time.time()
print(f"{round(tok-tik,1)} s")
print(f"answer:{r}")
```

```python
tik = time.time()
r = test_oss(QUESTION, 'medium')
tok = time.time()
print(f"{round(tok-tik,1)} s")
print(f"answer:{r}")
```

```python
tik = time.time()
r = test_oss(QUESTION, 'high')
tok = time.time()
print(f"{round(tok-tik,1)} s")
print(f"answer:{r}")
```
