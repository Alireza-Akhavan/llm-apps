# استفاده از **LiteLLM** با Ollama برای نقش‌های مختلف پزشکی

در این نوت‌بوک یاد می‌گیریم چطور با یک **System Prompt انگلیسی** به مدل بگوییم که همیشه **فارسی جواب بدهد**،
و همان سؤال را با سه نقش (Role) متفاوت بپرسیم تا ببینیم پاسخ‌ها چقدر فرق می‌کند.

```python
# ایمپورت کتابخانه‌های موردنیاز
import litellm
from litellm import completion
from IPython.display import display, Markdown

# این تابع هم System Prompt می‌گیرد و هم پرامپت اصلی کاربر (User Prompt)
# System Prompt به انگلیسی نوشته می‌شود، اما به مدل می‌گوییم که پاسخ را فارسی بدهد
def get_completion(prompt, system_prompt=None, model="ollama/gemma3:4b", temperature=0.2):
    messages = []
    if system_prompt:
        messages.append({"role": "system", "content": system_prompt})
    messages.append({"role": "user", "content": prompt})

    response = completion(
        model=model,
        messages=messages,
        temperature=temperature,
    )

    return response.choices[0].message.content

# چون پاسخ‌ها فارسی (راست‌به‌چپ) هستند، از این تابع برای نمایش درست استفاده می‌کنیم
def rtl_print(text):
    display(Markdown(f'<span style="direction: rtl; display: block; text-align: right;">{text}</span>'))

def ltr_print(text):
    display(Markdown(text))
```

## تعریف System Prompt (انگلیسی) و سؤال اصلی (فارسی)

- **System Prompt** به انگلیسی نوشته شده و به مدل نقش (Role) می‌دهد و می‌گوید که **حتماً به فارسی پاسخ دهد**.
- **سؤال اصلی (User Prompt)** به فارسی پرسیده می‌شود:  
  «عوارض جانبی ایبوپروفن چیست؟» (معادل فارسی: *Explain the side effects of ibuprofen*)

در ادامه همین سؤال را با سه System Prompt متفاوت (سه نقش) امتحان می‌کنیم.

```python
# سؤال اصلی به فارسی (همان چیزی که می‌خواهیم از مدل بپرسیم)
question = "عوارض جانبی ایبوپروفن چیست؟"
```

### 👨‍⚕️ نقش اول: پزشک عمومی (General Doctor)

```python
# System Prompt انگلیسی: نقش پزشک عمومی + دستور پاسخ فارسی
system_prompt_doctor = (
    "You are a general practitioner (family doctor) talking to a patient. "
    "Explain things in simple, everyday language and give practical advice. "
    "Always respond in Persian (Farsi), regardless of the language of the question."
)

response = get_completion(question, system_prompt=system_prompt_doctor)
rtl_print(response)
```

### 👩‍⚖️ نقش دوم: فارماکولوژیست (Pharmacologist)

```python
# System Prompt انگلیسی: نقش فارماکولوژیست (توضیح مکانیسمی و تخصصی) + دستور پاسخ فارسی
system_prompt_pharmacologist = (
    "You are a pharmacologist explaining drug mechanisms to a medical colleague. "
    "Use precise pharmacological and mechanistic terminology (e.g. enzyme inhibition, receptor effects). "
    "Always respond in Persian (Farsi), regardless of the language of the question."
)

response = get_completion(question, system_prompt=system_prompt_pharmacologist)
rtl_print(response)
```

### 🧑‍💻 نقش سوم: نویسنده محتوای پزشکی برای بیماران (Medical Writer for Patients)

```python
# System Prompt انگلیسی: نقش نویسنده محتوای پزشکی برای عموم مردم + دستور پاسخ فارسی
system_prompt_writer = (
    "You are a medical writer creating patient-friendly educational content. "
    "Use very simple, short sentences that a non-medical person can easily understand, and avoid jargon. "
    "Always respond in Persian (Farsi), regardless of the language of the question."
)

response = get_completion(question, system_prompt=system_prompt_writer)
rtl_print(response)
```
