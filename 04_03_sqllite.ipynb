```python
import dotenv
dotenv.load_dotenv()
```

```python
from langchain.chat_models import init_chat_model
from langchain.agents import create_agent
from langgraph.checkpoint.sqlite import SqliteSaver
import sqlite3

llm = init_chat_model("gpt-5.2", model_provider="openai", temperature=0)

# اتصال به دیتابیس SQLite (فایل روی دیسک ذخیره می‌شود)
conn = sqlite3.connect("chat_memory_v2.db", check_same_thread=False)
checkpointer = SqliteSaver(conn)

# ساخت agent با حافظه‌ی دائمی روی SQLite
agent = create_agent(
    model=llm,
    tools=[],  # بدون tool — فقط conversation
    checkpointer=checkpointer,
    system_prompt="You are a helpful assistant."
)

config = {"configurable": {"thread_id": "session-1"}}

# مثال استفاده
response = agent.invoke(
    {"messages": [{"role": "user", "content": "اسم من چیه؟"}]},
    config=config
)
print(response["messages"][-1].content)
```

```python
from langchain.chat_models import init_chat_model
from langchain.agents import create_agent
from langgraph.checkpoint.sqlite import SqliteSaver
import sqlite3

llm = init_chat_model("gpt-5.2", model_provider="openai", temperature=0)

# اتصال به دیتابیس SQLite (فایل روی دیسک ذخیره می‌شود)
conn = sqlite3.connect("chat_memory.db", check_same_thread=False)
checkpointer = SqliteSaver(conn)

# ساخت agent با حافظه‌ی دائمی روی SQLite
agent = create_agent(
    model=llm,
    tools=[],  # بدون tool — فقط conversation
    checkpointer=checkpointer,
    system_prompt="You are a helpful assistant."
)

config = {"configurable": {"thread_id": "session-1"}}

# مثال استفاده
response = agent.invoke(
    {"messages": [{"role": "user", "content": "سلام، اسمم علی است."}]},
    config=config
)
print(response["messages"][-1].content)
```

```python
from langgraph.checkpoint.sqlite import SqliteSaver

with SqliteSaver.from_conn_string("chat_memory.db") as checkpointer:
    agent = create_agent(
        model=llm,
        tools=[],
        checkpointer=checkpointer,
        system_prompt="You are a helpful assistant."
    )
    response = agent.invoke(
        {"messages": [{"role": "user", "content": "سلام"}]},
        config=config
    )
```
